### 1. Stack Data structure.
Stack = LIFO / FILO data structure. Last In First Out/ First In Last Out.  
        Stores objects into a sort of 'vertical tower'.
        Push() to add to the top.
        Pop() to remove from the top.

### Implementing stack using list

```python
stack = []

def push():
    if len(stack) == n:
        print('Stack is full')
    else:
        element = input('Enter an element:')
        stack.append(element)
        print(stack)

def pop_element():
    if not stack:
        print('stack is empty')
    else:
        e = stack.pop()
        print('removed element', e)
        print(stack)

n = int(input('Enter the limit of the stack'))
while True:
    print('select the operation: 1.Push 2.Pop 3.Exit')
    choice = int(input())
    if choice == 1:
        push()
    
    elif choice == 2:
        pop_element()

    elif choice == 3:
        break

    else:
        print('Enter a valid operation')
```

### Implementing stack using collections and Queue Modules.

Collections class also uses the same append and pop methods as list.

```python
import collections

stack = collections.deque()
print(stack)

stack.append(33)
stack.append(55)
stack.append(77)

stack
```

```python
print(stack.pop())
print(stack.pop())
print(stack.pop())

# The stack is empty now
print(stack.pop())
```

Using Queues.

Queues use put() and get() methods instead of append() and pop().

```python
import queue
stack = queue.LifoQueue(3) # Queue only takes 3 elements.

stack.put(33)
stack.put(55)
stack.put(77)
print(stack.queue)

stack.put(99, timeout = 1) # Timeout of 1 second so the program doesn't run forever waiting for slot to be empty.
```

```python
print(stack.get())
print(stack.get())
print(stack.get())

# The stack is empty now
print(stack.get(timeout = 1)) # Timeout of 1 second so the program doesn't run forever waiting for another element to get().
```

### 2. Queue Data Structure.

Queue - A linear Data structure where an element is inserted from one side and removed from another.  
&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; Is open on both sides.  
&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; Is a First In First Out/ Last In Last Out(FIFO/LILO) data structure.  
&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; End where elements are inserted - **back/rear/tail**.  
&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; End where elements are removed - **front/head**.  
        **Enqueue** - Process of adding element to a queue.  
        **Dequeue** - Process of removing element from a queue.

Queue data structure can be implemented using a lists or modules.

### Implementing Queue data structure using list.

```python
queue = []

queue.append(10)
queue.append(20)
queue.append(30)

queue
```

```python
print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))

# This functions as a queue because the elements are insterted in one end and removed removed from the other end.
```

```python
queue = []
def enqueue():
    element = input('Enter the element')
    queue.append(element)
    print('Element has been added to queue')

def dequeue():
    if not queue:
        print('Queue is empty')
    else:
        e = queue.pop(0)
        print('element removed:', e)

def display():
    print(queue)

while True:
    print('Select an Operation: 1. Enqueue 2. Dequeue 3. Show 4. Quit')
    choice = int(input())
    if choice == 1:
        enqueue()
    elif choice == 2:
        dequeue()
    elif choice == 3:
        display()
    elif choice == 4:
        break
    else:
        print('Enter a valid operation')
```
