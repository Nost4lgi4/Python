### 1. Stack Data structure.
Stack = LIFO / FILO data structure. Last In First Out/ First In Last Out.  
        Stores objects into a sort of 'vertical tower'.
        Push() to add to the top.
        Pop() to remove from the top.

### Stack Exercises:
1. Write a Python program to create a class representing a stack data structure. Include methods for pushing and popping elements.

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push_stack(self, element):
        self.stack.append(element)
        print(f'The added element is: {element}. The stack is now: {self.stack}',)
        return self.stack
    
    def pop_stack(self):
        if not self.stack:
            return 'Cannot pop from an empty stack'
        else:
            print('The popped element is:', self.stack.pop(), 'The stack is now:', self.stack)
        return self.stack

    def is_empty(self):
        if not self.stack:
            return ('Stack is empty')
        else:
            return ('Stack is not empty')
    
    def size(self):
        return len(self.stack)
    
    def peek(self):
        if not self.stack:
            return ('Stack is empty')
        else:
            return self.stack[-1]

stack1 = Stack()
stack1.push_stack(1)
stack1.push_stack(2)
stack1.push_stack(3)

print('The size of the stack is:', stack1.size())
print('The last element in the stack is:', stack1.peek())

stack2 = Stack()
stack2.push_stack(5)
stack2.push_stack(6)
stack2.push_stack(7)

stack2.pop_stack()
stack2.pop_stack()
stack2.pop_stack()
stack2.pop_stack()
```

2. Write a Python class for a Stack that maintains the current minimum element and provides a method to retrieve it in constant time.

```python
class Stack:
    def __init__(self):
        self.stack = []

    def get_min(self):
        return min(self.stack)

    def push_stack(self, element):
        self.stack.append(element)
        return self.stack

    def pop_stack(self):
        self.stack.pop()
        return self.stack

    def peek(self):
        return self.stack[-1]
    
stack1 = Stack()

stack1.push_stack(11)
stack1.push_stack(33)
stack1.push_stack(55)

stack1.get_min()
```

### 2. Queue Data Structure.

Queue - A linear Data structure where an element is inserted from one side and removed from another.  
&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; Is open on both sides.  
&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; Is a First In First Out/ Last In Last Out(FIFO/LILO) data structure.  
&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; End where elements are inserted - **back/rear/tail**.  
&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; End where elements are removed - **front/head**.  
        **Enqueue** - Process of adding element to a queue.  
        **Dequeue** - Process of removing element from a queue.

Queue data structure can be implemented using a lists or modules.

```python
queue = []
def enqueue():
    element = input('Enter the element')
    queue.append(element)
    print('Element has been added to queue')

def dequeue():
    if not queue:
        print('Queue is empty')
    else:
        e = queue.pop(0)
        print('element removed:', e)

def display():
    print(queue)

while True:
    print('Select an Operation: 1. Enqueue 2. Dequeue 3. Show 4. Quit')
    choice = int(input())
    if choice == 1:
        enqueue()
    elif choice == 2:
        dequeue()
    elif choice == 3:
        display()
    elif choice == 4:
        break
    else:
        print('Enter a valid operation')
```

### Linked List Data Structure.

Linked list - Linear Data structure made up of a chain of nodes in which each node contains a data field and link or reference.




Stores the reference of the next node.



##### Advantages of a linked list:

1. It's a dynamic data structure.
2. Insertion and deletion is easy.
3. Can be used to implement stack, queue, and graph.

##### Disadvantages of a linked list:

1. Needs extra memory.
2. Random access not possible.

Types of Linked lists:
1. Singly linked list - The nodes of the list only contain one link.
2. Doubly linked list - the nodes of the list contain two links.
3. Circular linked list.

I. Singly linked list.



Operations performed by a singly linked list:
1. Add.
2. Delete.
3. Traversal.

Steps to add elements to the beginning of a linked list:
1. Create a node.
2. Change new_node next = head.
3. head -> new node.



Steps to add elements to the end of a linked list:
1. Create a node.
2. Go to last node.
3. Reference of last node = new node.



Steps to add elements between two nodes of a linked list:
1. Create a node.
2. Go to node just before required position of new node.
3. x.ref = new node
   z.ref = y.



### Linked list in python.



```python
# Traversing a linked list.

class Node:
    def __init__(self, data):
        self.data = data
        self.ref = None

class LinkedList: # Traversing the linked list
    def __init__(self):
        self.head = None

    def print_LL(self):
        if self.head is None:
            print('Linked list is empty')
        else:
            n = self.head
            while n is not None:
                print(n.data, '--->', end = " ")
                n = n.ref

    def add_begin(self, data): # Add to the beginning of the linked list
        new_node = Node(data)
        new_node.ref = self.head
        self.head = new_node
        
    def add_end(self, data): # Add to the end of the linked list
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
        else:
            n = self.head
            while n.ref is not None:
                n = n.ref
            n.ref = new_node

ll1 = LinkedList()
ll1.add_begin(33)
ll1.add_end(55)
ll1.add_begin(11)
ll1.print_LL()
```
